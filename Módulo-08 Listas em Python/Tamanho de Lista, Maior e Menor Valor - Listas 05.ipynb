## Maior e Menor Valor

maior = max(lista)

menor = min(lista)

```python
vendas = [1000, 1500, 15000, 270, 900, 100, 1200]
```

- Qual o item mais vendido?
- Qual o item menos vendido?

```python
print(f'A maior venda encontrada na lista é: {max(vendas)}')
print(f'A menor venda encontrada na lista é: {min(vendas)}')
```

```python
# Para identificarmos o produto mais vendido e menos, podemos nos basear no índice da lista de vendas para pegar o nome na lista de produtos.

maior_venda = max(vendas)
produto_mais_vendido = vendas.index(maior_venda) # retorna o índice da lista de produto
print(f'O produto mais vendido foi {produtos[produto_mais_vendido]}')

menor_venda = min(vendas)
produto_menos_vendido = vendas.index(menor_venda)
print(f'O produto mais vendido foi {produtos[produto_menos_vendido]}')
```
