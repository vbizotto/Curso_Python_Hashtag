## 1. Faturamento do Melhor e do Pior Mês do Ano

Qual foi o valor de vendas do melhor mês do Ano?
E valor do pior mês do ano?

```python
meses = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
vendas_1sem = [25000, 29000, 22200, 17750, 15870, 19900]
vendas_2sem = [19850, 20120, 17540, 15555, 49051, 9650]

vendas_anual = vendas_1sem
vendas_anual += vendas_2sem
print(vendas_anual)

maior_venda = max(vendas_anual)
print(f'A maior venda foi de: {maior_venda}')

menor_venda = min(vendas_anual)
print(f'A menor venda foi de: {menor_venda}')

indice_mes_maior_venda = vendas_anual.index(maior_venda)
print(f'O mês da maior venda foi: {meses[indice_mes_maior_venda]}')

indice_mes_menor_venda = vendas_anual.index(menor_venda)
print(f'O mês da maior venda foi: {meses[indice_mes_menor_venda]}')
```

## 2. Continuação

Agora relacione as duas listas para printar 'O melhor mês do ano foi {} com {} vendas' e o mesmo para o pior mês do ano.

Calcule também o faturamento total do Ano e quanto que o melhor mês representou do faturamento total.

Obs: Para o faturamento total, pode usar a função sum(lista) que soma todos os itens de uma lista

```python
mes_maior_venda = meses[indice_mes_maior_venda]
mes_menor_venda = meses[indice_mes_menor_venda]
print(f'O melhor mês do ano foi {mes_maior_venda} com {maior_venda} vendas')
print(f'O pior mês do ano foi {mes_menor_venda} com {menor_venda} vendas')
```
