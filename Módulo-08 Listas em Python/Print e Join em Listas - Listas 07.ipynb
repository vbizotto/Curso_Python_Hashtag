# Print de Listas

2 Opções:
- print "normal"
- método join -> texto.join(lista)

```python
produtos = ['apple tv', 'mac', 'iphone x', 'iphone 11', 'IPad', 'apple watch', 'mac book', 'airpods']
print(produtos)

# Método join extrai os elementos de uma lista, transformando-a em uma string separando os elementos com um separador desejado

minha_string = (','.join(produtos))
print(minha_string)

print(','.join(produtos))
print('\n'.join(produtos)) # separa cada elemento com <ENTER>
```

Lembrando do método split de strings:

lista = texto.split(separador)

```python
produtos = 'apple tv, mac, iphone x, iphone 11, IPad, apple watch, mac book, airpods'

lista_produtos = produtos.split(',') # cria uma lista a partir de uma string com os elementos separados por um caracter específico

print(lista_produtos)
```
