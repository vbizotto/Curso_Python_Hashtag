### Forma Básica

```python
precos = "Jan: 25, Fev: 27, Mar: 29"

```

### Posição Inicial e Final

```python
preco_jan = precos[5:7]
print (preco_jan)

preco_fev = precos[14:16]
print(preco_fev)

preco_mar = precos[-2:]
print(preco_mar)
```

### Posição Inicial e Final com Step

```python
codigo = "1.2.3.4,5,1,2.3.4,7.9"
```

```python
pedaco_cod = codigo[:]
print(pedaco_cod)

pedaco_cod_step = codigo[::2]
print(pedaco_cod_step) # pega o número do início ao fim, saltando de 2 em 2

inverso = codigo[::-2]
print(inverso)
```
