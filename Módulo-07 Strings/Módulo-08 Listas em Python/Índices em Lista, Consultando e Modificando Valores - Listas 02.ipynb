# Listas em Python

## Estrutura:

lista = [valor, valor, valor, valor, ...]

lista[i] -> é o valor de índice i da lista. <br>
Obs: Lembre que no python o índice começa em 0, então o primeiro item de uma lista é o item lista[0]

Para substituir um valor de uma lista você pode fazer:<br>
lista[i] = novo_valor

Listas de Produtos de uma Loja:

```python
produtos = ['tv', 'celular', 'mouse', 'teclado', 'tablet']
print(produtos[1])
```

Lista de Unidades Vendidas de cada Produto da Loja

```python
vendas = [1000, 1500, 350, 270, 900]
print(vendas[1])
```

### Nesse caso, as listas funcionam da seguinte forma:

```python
vendas[1] = 2000
print('As vendas do produto {}, foram de {} unidades'.format(produtos[1], vendas[1]))
```

```python
texto = 'lira@gmail.com'
print(texto[1])
```

```python

# Embora possamos encarar a string como uma lista de caracteres, nós não podemos alterar o elemento desta lista, veja o erro que ocorre abaixo

texto[1] = 'a'
```

```python
# Para o tipo de situação acima podemos utillizar o método replace
print(texto)
texto = texto.replace('i','a')
print(texto)
```
