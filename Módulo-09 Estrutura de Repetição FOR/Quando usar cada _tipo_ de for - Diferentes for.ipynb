```python
produtos = ["iphone", "ipad", "airpod", "macbook"]
precos = [7000, 10000, 2500, 14000]
```

### For item in lista

```python
# preco com imposto
for preco in precos:
    print(preco * 1.10)
```

### For i in range

```python
# preco de cada produto
for i in range(len(produtos)):
    print(f'O preço do produto {produtos[i]} é {precos[i]}')
```

### For item in lista com enumerate

```python
# preco de cada produto com imposto
for i, produto in enumerate(produtos):
    print(f'O preço do produto {produto} com imposto é {precos[i] * 1.10}')
```
