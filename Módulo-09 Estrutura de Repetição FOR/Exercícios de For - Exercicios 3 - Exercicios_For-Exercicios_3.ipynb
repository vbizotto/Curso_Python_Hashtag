## 1. Calculando % de uma lista

Faremos algo parecido com "filtrar" uma lista. Mais pra frente no curso aprenderemos outras formas de fazer isso, mas com o nosso conhecimentoa atual já conseguimos resolver o desafio.

Digamos que a gente tenha uma lista de vendedores e ao invés de saber todos os vendedores que bateram a meta, eu quero conseguir calcular o % de vendedores que bateram a meta. Ou seja, se temos 10 vendedores e 3 bateram a meta, temos 30% dos vendedores que bateram a meta.

- Vamos resolver de 2 formas:
    1. Criando uma lista auxiliar apenas com os vendedores que bateram a meta
    2. Fazendo o cálculo diretamente na lista que já temos

```python
# Resolvendo criando uma lista auxiliar
meta = 10000
qtde_func_bateram_meta = 0
metas_batidas = []

vendas = [
    ['João', 15000],
    ['Julia', 27000],
    ['Marcus', 9900],
    ['Maria', 3750],
    ['Ana', 10300],
    ['Alon', 7870],
]
#seu código aqui
for venda in vendas:
    if venda[1] >= meta:
       metas_batidas.append(venda)

perc_func_bateram_meta = (len(metas_batidas) / len(vendas)) 
print(f'O percentual de vendedores que bateram a meta foi de {perc_func_bateram_meta:.2%}') 

```

```python
# Resolvendo sem criar uma lista auxiliar
meta = 10000
qtde_func_bateram_meta = 0

vendas = [
    ['João', 15000],
    ['Julia', 27000],
    ['Marcus', 9900],
    ['Maria', 3750],
    ['Ana', 10300],
    ['Alon', 7870],
]
#seu código aqui
for venda in vendas:
    if venda[1] >= meta:
       qtde_func_bateram_meta += 1

perc_func_bateram_meta = (qtde_func_bateram_meta / len(vendas)) 
print(f'Linhas de vendas = {len(vendas)}')
print(f'Quantidade de funcionários que bateram a meta {qtde_func_bateram_meta}')
print(f'O percentual de vendedores que bateram a meta foi de {perc_func_bateram_meta:.2%}') 
```
