# Adicionar, Remover e Modificar Itens no Dicionário

### Estrutura:

- Adicionar itens:

```python
lucro_1tri = {'janeiro': 100000, 'fevereiro': 120000, 'março': 90000}
lucro_2tri = {'abril': 88000, 'maio': 89000, 'junho': 120000}

#adicionando 1 item
lucro_2tri['julho'] = 1000
print(lucro_2tri)

#adicionando vários itens ou um dicionário a outro
lucro_3tri = {'jul': 3000, 'ago': 4000,'set':5000}
print(lucro_3tri)

#adicionando um item já existente (manualmente ou pelo update)
lucro_3tri['jul'] = 80000
print(f'Incluindo um item "jul" já existente, ele altera o valor da chave: {lucro_3tri}')
lucro_3tri.update({'jul': 6000})
print(f'Após realizar o updte de "jul" para 600: {lucro_3tri}')

#criando um novo dicionário adicionado outros dois dicionários já existentes
lucro_1sem = {}
lucro_1sem.update(lucro_1tri)
lucro_1sem.update(lucro_2tri)

print(lucro_1sem)
```

- Modificar itens:

Da mesma forma que adicionamos 1 valor, caso essa chave já exista o item é apenas modificado.

dicionario[chave] = valor

Vamos modificar o lucro de fevereiro:<br>
(Lembrando: caso o item não exista, ele vai criar o item no dicionário)

```python
lucro_jul = 85000
lucro_1sem['julho'] = lucro_jul
print(lucro_1sem)
```

- Remover itens:

del dicionario[chave]

ou então

valor = dicionario.pop(chave)

mas cuidado com:

del dicionario<br>    ->    que é diferente de dicionario.clear()

```python
#removendo elemento com a método pop
#Obs.: A utilização do pop força a atualização de uma variável com o valor excluído.
lucro_julho = lucro_1sem.pop('julho')
print (lucro_1sem)
print(lucro_julho)
```

```python
#removendo o mês de julho

del lucro_1sem['julho']
print(lucro_1sem)

```

```python
 #limpando o dicionário, o método .clear não elimina o dicionário, ele esvazia.
lucro_1sem.clear()
print(lucro_1sem)
```
