#### 3. Faça um programa que leia e valide as seguintes informações (e para cada uma delas, continue pedindo a informação até o usuário inserir corretamente):
##### Nome: maior que 3 caracteres;
##### Idade: entre 0 e 150;
##### Salário: maior que zero;
##### Sexo: 'f' ou 'm';
##### Estado Civil: 's', 'c', 'v', 'd';

````python
dados = []
while True:
    nome = input('Informe um nome que contenha mais de três caracteres: ')
    if len(nome) < 3:
        print('Informe um nome com mais de três caracteres.')
        continue
    
    idade = int(input('Informe a idade (Deve estar entre 0 e 150): '))
    if (idade < 0 or idade > 150):
        print('Informe uma idade entre 0 e 150 anos.')
        continue
    
    salario = int(input('Informe o valor de salário(Deve ser maior que 0): '))
    if salario < 0:
        print('Informe um salário maior que zero.')
        continue
    
    sexo = input('Informe o sexo(M/F): ')
    if sexo not in ('M', 'F'):
        print('Sexo deve ser: Masculino oui Feminino.')
        continue

    estado_civil = input('Informe o estado civil(S, C, V, D): ') 
    if estado_civil not in ('S', 'C', 'V', 'D'):
        print('Estado civil deveser: Solteiro(a), Casado(a), Viúvo(a), Divivorciado(a)')
        continue

    dados.append([nome, idade, salario, sexo, estado_civil])    
    break

print(dados)

````

#### 4. Supondo que a população de um país A seja da ordem de 80000 habitantes com uma taxa anual de crescimento de 3% e que a população de B seja 200000 habitantes com uma taxa de crescimento de 1.5%. Faça um programa que calcule e escreva o número de anos necessários para que a população do país A ultrapasse ou iguale a população do país B, mantidas as taxas de crescimento.

````python
populacao_a = 80000
populacao_b = 200000
taxa_a = 0.03
taxa_b = 0.015
anos = 0

while True:
    anos += 1
    populacao_a += (populacao_a * taxa_a)
    populacao_b += (populacao_b * taxa_b)

    print(f'Em {anos} anos a população do país A é de: {populacao_a:.2}, enquanto que a população do país B é de: {populacao_b:.2}')
    if populacao_a > populacao_b:
        break

print (f'O pais A ultrapassará a população do país B em {anos} anos')
````

#### 5. Altere o programa anterior permitindo ao usuário informar as populações e as taxas de crescimento iniciais. Valide a entrada e permita repetir a operação.

#### 8. Faça um programa que consiga categorizar a idade das equipes de uma empresa. Faça um programa que peça para n pessoas a sua idade, ao final o programa devera verificar se a média de idade da equipe varia entre 0 e 25 (jovem) ,26 e 60 (sênior) e maior que 60 (idosa); e então, dizer se a equipe é jovem, sênior ou idosa, conforme a média calculada.

````python
idades = []

while True:
    idade = int(input('Informe a idade dos membros da equipe(Para interromper informe 0)'))
    if idade == 0:
        break
    idades.append(idade)
            
print(f'A soma das idades é: {sum(idades)} e a quantidade de idades é {len(idades)}')
media_idades = sum(idades) / len(idades)
print(f'Média de idades é: {media_idades}')

if (media_idades >= 0 and media_idades <= 25):
    print('A equipe é jovem.')
elif (media_idades >= 26 and media_idades <= 60):
    print ('A equipe é sênior.') 
else:
    print ('A equipe é idosa')       
````

#### 9. Numa eleição existem três candidatos. Faça um programa que peça o número total de eleitores. Peça para cada eleitor votar e ao final mostrar o número de votos de cada candidato.

````python
candidatos = ['candidato 1', 'candidato 2', 'candidato 3', 'Nulos']
votos = [0,0,0,0]
eleitores = int(input('informe o número total de eleitores: '))

votantes = 0

while votantes < eleitores:
    candidato = input('Informe o nome do seu candidato: ')
    if candidato in candidatos:
        if candidato == candidatos[0]:
            votos[0] += 1
        elif candidato == candidatos[1]:
            votos[1] += 1
        elif candidato == candidatos[2]:
            votos[2] += 1       
    else:
        votos[3] += 1
    
    votantes += 1    

print('Resumo da votação')    
print(candidatos)
print(votos)

for i in range(len(candidatos)):
    print(f'O candidato {candidatos[i]} recebeu {votos[i]} votos')
````

#### 10. Faça um programa que calcule o valor total investido por um colecionador em sua coleção de CDs e o valor médio gasto em cada um deles. O usuário deverá informar a quantidade de CDs e o valor para em cada um.

````python
cds = []
valores_cd = []

while True:
    cd = input('Informe o título do CD: ')
    if not cd:
        break
    valor = float(input('Informe o valor do CD: '))

    cds.append(cd)
    valores_cd.append(valor)

print(cds)
print(valores_cd)

print (f'A coleção de CDs possui {len(cds)} com um valor total de R$ {sum(valores_cd)}')
print (f'O valor médio de cada CD é: {sum(valores_cd) / len(cds):.2f}') 
````

#### 12. Um funcionário de uma empresa recebe aumento salarial anualmente: Sabe-se que:

Em 1996 recebeu aumento de 1,5% sobre seu salário inicial;
A partir de 1997 (inclusive), os aumentos salariais sempre correspondem ao dobro do percentual do ano anterior. Faça um programa que determine o salário desse funcionário em 2003. 
</pre>

````python
salario_inicial = 1000
percentual = 1.5
ano = 1995
print(f'O salário do funcionário no ano de {ano} foi de {salario_inicial}')
novo_salario = salario_inicial
ano += 1
while ano <= 2003:
    print(f'{novo_salario} ---> {percentual}' )
    novo_salario = (novo_salario * (1 + (percentual / 100)))
    print(f'O salário do funcionário no ano de {ano} foi de {novo_salario:.2f}')
    percentual = (percentual * 2)
   
    ano += 1
````

#### 13. O cardápio de uma lanchonete é o seguinte:
<pre>
Especificação   Código  Preço
Cachorro Quente 100     R$ 1,20
Bauru Simples   101     R$ 1,30
Bauru com ovo   102     R$ 1,50
Hambúrguer      103     R$ 1,20
Cheeseburguer   104     R$ 1,30
Refrigerante    105     R$ 1,00
Faça um programa que leia o código dos itens pedidos e as quantidades desejadas. Calcule e mostre o valor a ser pago por item (preço * quantidade) e o total geral do pedido. Considere que o cliente deve informar quando o pedido deve ser encerrado.
</pre>

````python
produtos = ['Cachorro Quente', 'Bauru Simples','Bauru com Ovo', 'Hamburguer','Cheeseburguer', 'Refrigerante']
cod_produtos = [100,101,102,103,104,105]
precos = [1.2,1.3,1.5,1.2,1.3,1.0]
pedidos = []
total_pedido = 0

while True:
    codigo = int(input('Informe o código do produto para seu pedido(Informe 0 para encerrar o pedido): '))
    if codigo == 0:
        break
    if codigo not in cod_produtos:
        print('Código produto inexistente')
    qtde = int(input('informe a quantidade do produto: '))  
    ind = cod_produtos.index(codigo)
    preco = precos[ind]
    total_item = (qtde * preco)
   
    pedidos.append([codigo,produtos[ind],qtde,preco,total_item])

for i, pedido in enumerate(pedidos):
    print(f'{i} - {pedidos[i][0]} - {pedidos[i][1]} - {pedidos[i][2]} - {pedidos[i][3]} - {pedidos[i][4]}')
    
total_pedido =  sum(pedido[4] for pedido in pedidos)
print(f'Total pedido = R$  {total_pedido}')
````

#### 14. Uma grande emissora de televisão quer fazer uma enquete entre os seus telespectadores para saber qual o melhor jogador após cada jogo. Para isto, faz-se necessário o desenvolvimento de um programa, que será utilizado pelas telefonistas, para a computação dos votos. Sua equipe foi contratada para desenvolver este programa, utilizando a linguagem de programação Python. Para computar cada voto, a telefonista digitará um número, entre 1 e 23, correspondente ao número da camisa do jogador. Um número de jogador igual zero, indica que a votação foi encerrada. Se um número inválido for digitado, o programa deve ignorá-lo, mostrando uma breve mensagem de aviso, e voltando a pedir outro número. Após o final da votação, o programa deverá exibir:
<pre>a. O total de votos computados;
b. Os números e respectivos votos de todos os jogadores que receberam votos;
c. O percentual de votos de cada um destes jogadores;
d. O número do jogador escolhido como o melhor jogador da partida, juntamente com o número de votos e o percentual de votos dados a ele.
        Observe que os votos inválidos e o zero final não devem ser computados como votos. O resultado aparece ordenado pelo número do jogador. O programa deve fazer uso de arrays. O programa deverá executar o cálculo do percentual de cada jogador através de uma função. Esta função receberá dois parâmetros: o número de votos de um jogador e o total de votos. A função calculará o percentual e retornará o valor calculado. Abaixo segue uma tela de exemplo. O disposição das informações deve ser o mais próxima possível ao exemplo. Os dados são fictícios e podem mudar a cada execução do programa. Ao final, o programa deve ainda gravar os dados referentes ao resultado da votação em um arquivo texto no disco, obedecendo a mesma disposição apresentada na tela.</pre>
```
Enquete: Quem foi o melhor jogador?

Número do jogador (0=fim): 9
Número do jogador (0=fim): 10
Número do jogador (0=fim): 9
Número do jogador (0=fim): 10
Número do jogador (0=fim): 11
Número do jogador (0=fim): 10
Número do jogador (0=fim): 50
Informe um valor entre 1 e 23 ou 0 para sair!
Número do jogador (0=fim): 9
Número do jogador (0=fim): 9
Número do jogador (0=fim): 0

Resultado da votação:

Foram computados 8 votos.

Jogador Votos           %
9               4               50,0%
10              3               37,5%
11              1               12,5%
``` 

````python
# Como é um número limitado de jogadores 1 à 23 vamos adotar o ídice da lista como sendo o voto do jogador, ignorando o índice 0
votos=[]
for i in range(23):
    votos.append(0)

while True:
    jogador = input('Informe um número de jogador de 1 a 23 que foi o melhor da partida. - Para encerrar a votação informe 0: ')
    if jogador:
        jogador = int(jogador)
        if jogador == 0:
            print('Votação encerrada.')
            break
        elif jogador > 23:
            print('O número de jogador deve ser um número de 1 a 23 ou 0 para encerrar a votação. Tente novamente')
            continue

        votos[jogador] += 1

print(f'Foram computados {sum(votos)} votos.')

print (f'Jogador  Votos    %')
print(votos)

for i, qtde_votos in enumerate(votos):
    if i > 0:
        print(f'  {i}          {qtde_votos}     {qtde_votos / sum(votos):.2%}')

print(f'O melhor jogador da partida foi {votos.index(max(votos))} com {max(votos)} votos que correspondem à {max(votos) / sum(votos):.2%}')
````
